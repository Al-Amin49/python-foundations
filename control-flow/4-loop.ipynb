```python
# A for loop repeats a block of code for each item in a collection.

fruits=["Apples","Bananas","Cherries"]
for fruit in fruits:
    print(fruit)
```

```python
# start and stop
for i in range(2, 6):
    print(i)  # notice that 6 is not included

# step value
# 2, 10, 2(step value)
for j in range(2, 10, 2):
    print("\nstep value",j)  # notice that 10 is not included
```

```python
#counting backwards
for k in range(10, 0, -1):
    print("counting backwards",k)  # notice that 0 is not included
```

```python
#loop through a string
for char in "Hello, World!":
    print("loop through a string", char)
#loop through a list
fruits=["Apples","Bananas","Cherries"]
for fruit in fruits:
    print("loop through a list", fruit)
#loop through a tuple
colors=("red", "green", "blue")
for color in colors:
    print("loop through a tuple", color)
#loop through a dictionary
student={"name":"John", "age":25, "city":"New York"}
for key in student:
    print("loop through a dictionary", key, student[key])
#loop through a set (set have no guaranteed order, so the order of items may vary each time you run the code)
numbers={1, 2, 3, 4, 5}
for n in numbers:
    print("loop through a set", n)
```

```python
fruits = ["Apple", "Banana", "Orange"]

for fruit in fruits:
    print(fruit)
# sometimes you need  the index
for index in range(len(fruits)):
    print("index", index, "fruit", fruits[index])
```

```python
# instead of range(len(fruits)), you can use enumerate() to get both the index and the value in a more Pythonic way:
for index, fruit in enumerate(fruits):
    print("index", index, "fruit", fruit)
```

```python
# loop control
#  break example
for i in range(10):
    if i == 5:
        break
    print("break example", i)

# continue example
for i in range(10):
    if i == 5:
        continue
    print("continue example", i)

# pass
# when you haven't written the code yet, but you want to have a placeholder for it, you can use the pass statement. It does nothing and is used as a placeholder.
for i in range(5):
    pass
```

```python
# else with for
# for → else runs if the loop finishes normally.
# break → else is skipped.

for i in range(5):
    print("else with for", i)
else:
    print("for loop completed without break")
```

```python
# nested loops
#Using Nested Loops – O(N²) Time and O(1) Space
for i in range(3):
    for j in range(2):
        print(i, j)
```

```python
for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
        
```

```python
for i in range(1, 4): # for column  
    for j in range(1, 4): # for row
        print(i * j, end=" ")
    print() # for a new line after each row
```

```python
# 1 to 20 multiplication table
for i in range(1, 21):
    for j in range(1, 11):
        print(i, 'x', j, '=', i*j)
        print(end=" ")
```

```python
for i in range(1, 4):
    for j in range(1, 4):
        if i == j:
            print("*", end=" ")
        else:
            print("-", end=" ")
    print()
```

```python
for i in range(1, 5):
    for j in range(i):
        print(j + 1, end=" ")
    print()
```

```python
# Pattern
# Pattern list --> https://www.geeksforgeeks.org/dsa/pattern-printing-problems/
N=3
for row in range(N):
    for col in range(row+1):
        print("*", end=" ")
    print()
```

```python
# nested list traversing using nested loops

bajarerList=[['Potato', 'Tomato', 'Onion'], [20, 30, 40], ['Soap', 12.5, 30]]

for item in bajarerList:
    for sub_item in item:
        print(sub_item, end=" ")
    print()  # for a new line after each sublist
```

```python
number=12345
count=0
while number>0:
    number //= 10
    count += 1
print("Number of digits:", count)
```

```python
number=12345
reverse=0
while number>0:
    digit=number%10
    reverse=reverse*10+digit
    number//=10
print("Reversed number:", reverse)
```
