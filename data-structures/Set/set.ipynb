
**A set is a collection of values where every value is unique. If you try to add a duplicate, it simply gets ignored. Sets are perfect when you care about what is present, but not about order or how many times it appears.**

- Every element is unique
- Elements are unordered
- Sets are mutable
- Set elements must be hashable
- Sets do not support indexing like lists/tuples

# Python: List vs Tuple vs Set

## Quick Comparison

| Feature | List | Tuple | Set |
|---|---|---|---|
| Syntax | `[]` | `()` | `{}` |
| Ordered | ✅ Yes | ✅ Yes | ❌ No guaranteed order |
| Mutable | ✅ Yes | ❌ No | ✅ Yes |
| Duplicates | ✅ Allowed | ✅ Allowed | ❌ Not allowed |
| Indexing | ✅ Yes | ✅ Yes | ❌ No |
| Main Purpose | Changeable collection | Fixed collection | Unique values |




**Summary**
- List = changeable + ordered
- Tuple = fixed + ordered
- Set = unique + unordered

```python
# set with numbers
numbers={10,20,30,40}
print(numbers)
# output:# {40, 10, 20, 30}
#The order may be different because a set is unordered.


# set automatically remove duplicates
num={1,2,2,3,4,4}
print(num)
```

```python
# creating a set

# set with string values
fruits={"apple", "mango", "banana"}

print(f"Fruits: {fruits}")

# set with mixed data types
mixed={"Ratul", 21, True, 9.5}
print(f"Mixed set: {mixed}")

# creating a set from a list, automatically removing duplicates

my_list=[1,2,3,4,5,3,4]

unique= set(my_list)
print(f"Unique elements from list {unique}")# 1, 2, 3, 4, 5}
```

```python
# Why sets don't allow duplicates

# The fundamental characteristic of a set in Python is that it only stores unique values. When you attempt to add an element that already exists. Python simply
# ignores the duplicate without raising an error or warning you. This makes sets incredibly efficient for ensuring data uniqueness.

# A practical use - removing duplicates from a list.


emails = ["a@gmail.com", "b@gmail.com", "a@gmail.com", "c@gmail.com"]
unique_emails = list(set(emails))
print(unique_emails)
# Output: ['a@gmail.com', 'b@gmail.com', 'c@gmail.com']
# (Note: The order may vary since sets are inherently unordered)
# This demonstrates a common use case: quickly filtering out duplicate entries from any iterable, like a list of email addresses. Converting to a set and then
# back to a list is a concise way to achieve this
```

```python
# counting unique words in a sentence.
sentence="Python is great and python is easy"

unique_words=set(sentence.split())
print(len(unique_words)) # 6
print(unique_words) # {'Python', 'is', 'easy', 'great', 'and', 'python'}
```

```python
# Accessing elements (Membership check)

# Since sets are unordered, direct access to elements using an index is not possible. Instead, you can efficiently verify if a value exists within a set using the in
# and not in operators.
fruits = {"apple", "mango", "banana"}
# This will raise a TypeError: 'set' object is not subscriptable
# print(fruits[0])

print("apple" in fruits) # Output: True
print("grapes" in fruits) # Output: False
print("kiwi" not in fruits) # Output: True
# This method is particularly powerful for scenarios requiring quick verification of an item's presence.
# Practical use — checking if a user has access

allowed_users = {"kabir", "Rajib", "alamin"}
user = input("Enter username: ")
if user in allowed_users:
    print("Access granted.")
else:
    print("Access denied.")
# Note: Membership checks (in) are exceptionally fast with sets, significantly outperforming similar operations on lists. This efficiency is a primary
# reason sets are preferred for lookups in real-world applications.
```

```python
# Adding and removing elements

# sets are mutable , meaning you can add new elements or remove existing ones after creation. Python provides several methods for these operations

# Adding elements
# Use add() for single elements, and update() for multiple elements from an iterable. Duplicates are silently ignored.
pythonfruits = {"apple", "mango"}
# add() — adds a single element
fruits.add("banana")
print(fruits) # {'apple', 'mango', 'banana'}
# update() — adds multiple elements (from any iterable)
fruits.update(["grapes", "kiwi"])
print(fruits) # {'apple', 'mango', 'banana', 'grapes', 'kiwi'}
# Adding a duplicate is silently ignored
fruits.add("apple")
print(fruits) # no change
```

```python
# removing elements
# remove, discard, pop, 

pythonfruits = {"apple", "mango", "banana", "kiwi"}
# remove() — removes the value, raises KeyError if not found
fruits.remove("mango")
print(fruits) # {'apple', 'banana', 'kiwi'}
# fruits.remove("grapes") # KeyError: 'grapes'
# discard() — removes if present, no error if not found (safer)
fruits.discard("banana") # works
fruits.discard("grapes") # silently does nothing
# pop() — removes and returns a random element
removed = fruits.pop()
print(removed) # any element
# clear() — removes everything
fruits.clear()
print(fruits) # set()
#Quick Tip: Use discard() when you are not sure whether the element exists. Use remove() only when you are sure — or when you want an error to catch a bug

```

```python
# set operations: Union and intersection

# sets are powerful tools for performing mathematical set operations. The two most common operations are union and intersection which allow you to combine or find common elemenets between sets

a={1,2,3,4}
b={3,4,5,6}

# using the | operator

print(a|b) # {1, 2, 3, 4, 5, 6}

# using the & operator
print(a&b)
```

```python
#Consider two sets representing students enrolled in different subjects. Set operations can easily identify students taking both or any of the subjects.

math_students = {"Karim", "Priya", "Al Amin", "Sana"}
science_students = {"Priya", "Kamran", "Arman"}
print("Students in BOTH Math and Science:")
print(math_students & science_students)
# Output: {'Priya', 'Karan'}

print("\nStudents in EITHER Math OR Science:")
print(math_students | science_students)
# Output: {'Rahul', 'Priya', 'Karan', 'Sana', 'Aman'}
#Note: Both the operator (|, &) and the method (.union(), .intersection()) achieve the same result. Choose the one that enhances readability for your specific use case. The methods can also take multiple iterables as arguments.
```

```python
# looping thorugh a set

for fruit in fruits:
    print(fruit)
```

```python
#While less common, you can also use enumerate, though it doesn't offer much practical benefit for sets given their unordered nature.
# Looping with enumerate (rarely needed for sets, but works)
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")
```

```python
items_list = ["apple", "mango", "apple"] # keeps order, keeps duplicates
items_tuple = ("apple", "mango", "apple") # keeps order, immutable
items_set = {"apple", "mango", "apple"} # unordered, removes duplicates

print(f"List: {items_list}") # Output: List: ['apple', 'mango', 'apple']
print(f"Tuple: {items_tuple}") # Output: Tuple: ('apple', 'mango', 'apple')
print(f"Set: {items_set}") # Output: Set: {'apple', 'mango'} (order may vary)
```
