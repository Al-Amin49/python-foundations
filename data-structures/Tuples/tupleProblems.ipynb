```python
# 5. Write a function get_stats(nums) that takes a tuple of numbers and returns a tuple containing the sum, average, minimum, and maximum. Unpack the returned tuple and print each value

# Unpacking = taking values from a tuple and putting them into separate variables.
nums = (3, 4, 5, 6, 7)

def get_stats(nums):
    total = sum(nums)
    average = total / len(nums)
    minimum = min(nums)
    maximum = max(nums)
    return (total, average, minimum, maximum)

# Call the function and unpack the returned tuple
total, average, minimum, maximum = get_stats(nums)

print("Sum:", total)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
# packing example
# Packing = putting multiple values into one variable.
def get_stats(*nums):  # packing: collects all args into a tuple called nums
    total = sum(nums)
    average = total / len(nums)
    minimum = min(nums)
    maximum = max(nums)
    return (total, average, minimum, maximum)

# Call with individual numbers, not a tuple
total, average, minimum, maximum = get_stats(3, 4, 5, 6, 7)

print("Sum:", total)
print("Average:", average)
print("Minimum:", minimum)
print("Maximum:", maximum)
print("Maximum:", maximum)
```
