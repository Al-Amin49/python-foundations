```python
#tuples
#A tuple is an ordered, immutable collection of python objects
#Once created, you cannot change its elements.
# its like bhai of a list , list is mutable 

""" 
Key priorities of a tuple

- Ordered --> items stay in the order you put them
- Immutable--> Cannot be modified after creation
- Allow duplicates --> the same value can appear more than once
- Can hold any data type --> integar, strings, boolean etc.
Tuples are used when you want to store data that should not change, such as cooridinates, RGB color values or database records when content should remain constatnt

"""
# creating tuple
numbers = (10, 20, 30, 40)
print(numbers)

```

### List vs Tuple


```python
# a tuple can contain different data types

person=("Al Amin", 25, 3.12, True)
print(person)
```

```python
# empty tuple
empty=()
print(type(empty))

#nested tuple
data=((1,2), (3,4), (5,6))
print(data)
```

```python
# single element tuple -- very important

x=(10)
print(type(x)) # type int
y=(10,)
print(type(y)) # tuple int
```

```python
# tuple indexing
numbers=(10,20,30,40,50)
print(numbers[0])
print(numbers[1])
# Negative indexing
print(numbers[-1])
print(numbers[-2])

# tuple slicing
print(numbers[1:4]) # (20, 30, 40) -- End index is excluded
print(numbers[:3]) # (10,20,30)
print(numbers[2:]) # (30, 40, 50)
print(numbers[:]) # all
```

```python
# slicing with step
numbers=(1,2,3,4,5,6)
print(numbers[::2]) # (1, 3, 5)
print(numbers[::-1]) (6, 5, 4, 3, 2, 1)
```

```python
 # Tuples are immutable 
# You cannot change directly an element

numbers= (10,20,30)
#numbers[0]=100 # TypeError: 'tuple' object does not support item assignment
```

```python
#  But you can reassign the entire tuple
numbers=(10,20,30)
# you can assign a completely new tuple
numbers=(100,200,300)

```

```python
# Tuple packing --> Many values → one variable
# Packing = putting multiple values into one variable.
person="Al Amin", 26, "Bangladesh"
print(person) # Here, three values are packed into one tuple:

# you can write also 
student = "Al Amin", 26, "CSE"
```

```python
# tuple unpacking
# One collection → many variables
# Unpacking = taking values from a tuple and putting them into separate variables.
person=("Al Amin", 26, "Bangladesh")
name, age, country=person
print(name)
print(age)
print(country)
```

```python
numbers = (10, 20, 30)

a, b, c = numbers
# a, b = numbers # this will not work
```

```python
# for unpacking multiple values
numbers = (10, 20, 30, 40, 50)

a, *others = numbers

print(a)
print(others)
```

```python
numbers = (10, 20, 30, 40, 50)

first, *middle, last = numbers

print(first)
print(middle)
print(last)
```

```python
# tuple methods
# count, index, len, min(), max(), sum()

numbers=(1,2,3,3,2,4)

print(numbers.count(2))
```

```python
# index 
numbers=(10,20,30,40)

print(numbers.index(30))
```

```python
numbers=(10,20,30,40)

print(len(numbers))
```

```python
#checking membership
numbers=(10,20,30)
print(20 in numbers)
```

```python
# loop through a tuple

numbers=(10,20,30,40)

for number in numbers:
    print(number)
```

```python
# tuple with enumerate (have to know)

names=("Rahim", "Karim", "Hasan")

for index, name in enumerate(names):
    print(index, name)
```

```python
# concetenating tuples
a=(1,2,3)
b=(4,5,6)
c=a+b
print(c)
```

```python
# Repeating tuples
numbers=(1,2)
print(numbers*3)
```

```python
# convert list
numbers=[1,2,3,4]

numbers_tuple=tuple(numbers)

print(numbers_tuple)
```

```python
# convert tuple -- List

numbers=(1,2,3,4)
numbers_list=list(numbers)
print(numbers)
```

```python
# tuple containing a list
data = (10, 20, [30, 40])
data[2].append(50)
print(data)
```

```python
# comparing tuples
a = (1, 2, 3)
b = (1, 2, 4)

print(a < b)

# sorting tuples
numbers = (5, 2, 8, 1, 3)

result = sorted(numbers)

print(result)
```

```python
# tuple as a dictionar key
locations = {
    (23.81, 90.41): "Dhaka",
    (24.89, 91.87): "Sylhet"
}
print(locations[(24.89, 91.87)])
```

```python
# why tuples are useful
# Use tuples when data should conceptually remain fixed.

# *args in tuple
def show(*args):
    print(type(args))
    print(args)

show("Ali", 25, "Bangladesh")
```
