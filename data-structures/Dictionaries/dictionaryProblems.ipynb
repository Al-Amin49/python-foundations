```python
"""
Q6: Student marks
Find:

Total marks
Average marks
Highest mark
Lowest mark

"""
marks = {
    "Bangla": 75,
    "English": 80,
    "Math": 90,
    "Physics": 85
}
total_sum=sum(marks.values())
average=total_sum/len(marks)
highest_mark=max(marks.values())
lowest_mark=min(marks.values())

print("Total marks:", total_sum)
print("Average marks:", average)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
```

```python
# q6 without using built-in functions
total_sum=0

for mark in marks.values():
    total_sum += mark
average=total_sum/len(marks)

highest_mark=0
for mark in marks.values():
    if mark> highest_mark:
        highest_mark=mark

lowest_mark = list(marks.values())[0]
for mark in marks.values():
    if mark < lowest_mark:
        lowest_mark = mark

print("Total marks:", total_sum)
print("Average marks:", average)
print("Highest mark:", highest_mark)
print("Lowest mark:", lowest_mark)
```

```python
"""
Count the frequency of each number in a list.
"""
numbers = [1, 2, 2, 3, 3, 3, 4, 4, 4, 4]
freq={}
for num in numbers:
    freq[num]=freq.get(num,0)+1
print(freq)
```

```python
# Count the frequency of each character in a string.
word = "programming"#input
freq={}
for char in word:
    freq[char]=freq.get(char,0)+1
print(freq)

# output: {'p': 1, 'r': 2, 'o': 1, 'g': 2, 'a': 1, 'm': 2, 'i': 1, 'n': 1}
```

```python
# Count the frequency of each word in a sentence.
sentence = "python is easy and python is powerful" # input
freq={}
for word in sentence.split():
    freq[word]=freq.get(word,0)+1
print(freq)

#output: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
```

```python
# Even and odd count in a list
numbers = [10, 15, 20, 25, 30, 35, 40] # Input
even_odd_count={"even": 0, "odd": 0}
for num in numbers:
    if num % 2 == 0:
        even_odd_count["even"] += 1
    else:
        even_odd_count["odd"] += 1
print(even_odd_count)

# output: {'even': 4, 'odd': 3}
```

```python
# unique values in a dictionary
data = {
    "a": 10,
    "b": 20,
    "c": 10,
    "d": 30,
    "e": 20
}
#without using built-in functions
unique_values = []
for value in data.values():
    if value not in unique_values:
        unique_values.append(value)
print(unique_values) # output should be [10, 20, 30]


# with built-in functions
unique_values = set(data.values())
print(unique_values) # output should be {10, 20, 30}
```

```python
"""
5. **Subject Performance Analysis:** Given a dictionary of marks for different subjects, loop over its values()
to calculate and print the total marks and the average mark obtained.

"""

marks={
  "English":90,
  "Science":67,
  "Mathematics":87,
  "History":56,
  "Islam":92
}
total=0
for value in marks.values():
  total+=value
print(total)
average=total/len(marks)
print(average)
```

```python
""" 
7. **Dictionary Merge Function:** Write a Python function named merge_dicts(d1, d2) that accepts two dictionaries (d1 and d2) as arguments and returns a new dictionary formed by merging them using the update() method. Ensure d1 remains unchanged.
"""

def merge_dicts(d1, d2):
    merged = d1.copy()
    merged.update(d2)
    return merged


d1 = {"name": "Al Amin", "age": 26}
d2 = {"city": "Sylhet", "country": "Bangladesh"}

result = merge_dicts(d1, d2)

print(result) # {'name': 'Al Amin', 'age': 26, 'city': 'Sylhet', 'country': 'Bangladesh'}
print(d1) # {'name': 'Al Amin', 'age': 26}

# if i give not copy then d1 will be changed after update
# requirement was to keep d1 unchanged 
d1.update(d2)
print(d1) # {'name': 'Al Amin', 'age': 26, 'city': 'Sylhet', 'country': 'Bangladesh'}
```

```python
 
""" 
8. **Subjects and Marks:** Create a dictionary of 6 subjects and their respective marks. Print the subject with the highest marks 
and the one with the lowest, using max() and min() functions alongside a lambda expression.
"""
subjects={
    "Math":96,
    "English":64,
    "Physics":51,
    "Chemistry":92,
    "Biology":75,
    "Computer":71
}
highest_marks_subject=max(subjects,key=lambda x:subjects[x])
lowest_marks_subject=min(subjects,key=lambda x:subjects[x])
print(f"The subject with the highest marks is {highest_marks_subject} with {subjects[highest_marks_subject]} marks")
print(f"The subject with the lowest marks is {lowest_marks_subject} with {subjects[lowest_marks_subject]} marks")
```

```python
""" 
10. **Student Marks Analysis:** Design a dictionary where each key is a student's name and the corresponding value is a list of their marks in 3 different subjects. Calculate and print the total marks and average marks for each student.
"""
studentMarks={
    "Al Amin":[84,40, 72],
    "Rahim":[90, 91, 54],
    "Haris":[65,98,73]
}

for name, marks in studentMarks.items():
    total=0
    for mark in marks:
        total+=mark
    print(name, total)

```

```python
"""
11. **Top 3 Subjects by Marks :** Given a dictionary of subjects and their marks, sort it by marks in descending
order. Then, print only the top 3 subjects with the highest marks.
"""
subjects={
    "Math":90,
    "English":93,
    "History":78,
    "Biology":81,
    "Islam":99
}

sorted_marks = sorted(
        subjects.items(),
        key=lambda item: item[1],
        reverse=True
    )

top3=sorted_marks[0:3]
print(top3)
```
