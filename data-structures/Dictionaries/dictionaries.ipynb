A dictionary is a collection of key–value pairs.
- Keys → must be immutable (string, number, tuple)
- Values → can be any data type
- dictionari has no indexing -- you have to access through key
- Dictionaries are mutable (can change after creation)
- A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
**As of Python version 3.7, dictionaries are ordered. In Python 3.6 and earlier, dictionaries are unordered.**

### Important dictionary methods
| Method         | Purpose                    |
| -------------- | -------------------------- |
| `get()`        | Safely get value           |
| `keys()`       | Get keys                   |
| `values()`     | Get values                 |
| `items()`      | Get key-value pairs        |
| `update()`     | Add/update multiple values |
| `pop()`        | Remove specific key        |
| `popitem()`    | Remove last item           |
| `clear()`      | Remove everything          |
| `copy()`       | Make shallow copy          |
| `setdefault()` | Get/add default value      |

```python
# dictionary comprehension

#normal way using loop
squares={}
for x in range(1, 5):
    squares[x]=x*x
print(squares) #{1: 1, 2: 4, 3: 9, 4: 16}

# dictionary complrehension -->same result in one line
squares={x:x*x for x in range(1, 5)}
print(squares) #{1: 1, 2: 4, 3: 9, 4: 16}

# dictionary comprehension with condition
subjects={"Math": 90, "Science": 85, "English": 88}
high_scores={key: value for key, value in subjects.items() if value > 85}
print(high_scores) # {'Math': 90, 'English': 88}
# double score
double_scores={key:value*2 for key, value in subjects.items()}
print(double_scores) # {'Math': 180, 'Science': 170, 'English
```

```python
# zip -- The zip() function in Python is a built-in tool used to combine multiple iterables (like lists, tuples, or strings) into a single iterator of tuples

# create dictionary from two lists
names=["Al Amin", "Akash", "Alom"]
ages=[26,27,25,28]

# use zip()

students=dict(zip(names, ages))
print(students) #{'Al Amin': 26, 'Akash': 27, 'Alom': 25}

# from 3 lists
names=["Al Amin", "Akash", "Alom"]
ages=[26,27,25]
departments=["CSE", "EEE", "BBA"]

for names, ages, departments in zip(names, ages, departments):
    print(f"Name: {names}, Age: {ages}, Department: {departments}")

# in list
names = ["Rahim", "Karim", "Hasan"]
marks = [80, 75]

print(list(zip(names, marks))) # [('Rahim', 80), ('Karim', 75)]
```

```python
# counting frequency using dictionary
numbers = [1, 2, 2, 3, 3, 3, 4, 7, 7]
freq={}
for number in numbers:
    if number in freq:
        freq[number]+=1
    else:
        freq[number]=1
print(freq)

# another way
frequency = {}

for number in numbers:
    frequency[number] = frequency.get(number, 0) + 1
print(frequency)
```

```python
""" 
1. Sorted the dictionary based on last marks(like 43, 60, 55, 105, 90)
2. Sorted the dictionary based on total marks
3. Sorted the dictionary based on average marks
4. Print the name of the student who has the highest marks
5. Print the name of the student who has the lowest marks
"""
# List in dictionary
marksByPeople={
    "Al Amin":[85, 90, 75, 80, 45, 63, 23, 43],
    "John Doe":[95, 85, 80, 90, 70, 65, 55, 60],
    "Jane Smith":[80, 85, 90, 75, 65, 70, 60, 55],
    "Philip": [70, 75, 80, 85, 90, 95, 100, 105],
    "Harry": [60, 65, 70, 75, 80, 85, 90]
}

#1
result= dict(sorted(marksByPeople.items(), key=lambda item: item[1][-1]))
print('sorted by last marks:', result)
#2
sorted_by_total = dict(sorted(marksByPeople.items(), key=lambda item: sum(item[1])))
print('sorted by total marks:', sorted_by_total)
#3
sorted_by_average= dict(sorted(marksByPeople.items(), key=lambda item: sum(item[1])/len(item[1])))
print('sorted by average marks:', sorted_by_average)
#4
highest_scorer = max(marksByPeople.items(), key=lambda item: sum(item[1]))
print("higest scorer:", highest_scorer)
```

```python
#without build in function sum, max and min
marksByPeople = {
    "Al Amin": [85, 90, 75, 80, 45, 63, 23, 43],
    "John Doe": [95, 85, 80, 90, 70, 65, 55, 60],
    "Jane Smith": [80, 85, 90, 75, 65, 70, 60, 55],
    "Philip": [70, 75, 80, 85, 90, 95, 100, 105],
    "Harry": [60, 65, 70, 75, 80, 85, 90]
}


# 1. Sort by last mark

result = dict(
    sorted(
        marksByPeople.items(),
        key=lambda item: item[1][-1]
    )
)

print("Sorted by last marks:", result)


# 2. Sort by total marks

totals = {}

for name, marks in marksByPeople.items():
    total = 0

    for mark in marks:
        total += mark

    totals[name] = total

sorted_by_total = dict(
    sorted(
        totals.items(),
        key=lambda item: item[1]
    )
)

print("Sorted by total marks:", sorted_by_total)


# 3. Sort by average marks

averages = {}

for name, marks in marksByPeople.items():
    total = 0

    for mark in marks:
        total += mark

    average = total / len(marks)

    averages[name] = average

sorted_by_average = dict(
    sorted(
        averages.items(),
        key=lambda item: item[1]
    )
)

print("Sorted by average marks:", sorted_by_average)


# 4. Highest scorer

highest_score = 0
highest_scorer = None

for name, marks in marksByPeople.items():

    total = 0

    for mark in marks:
        total += mark

    if total > highest_score:
        highest_score = total
        highest_scorer = name

print("Highest scorer:", highest_scorer)


# 5. Lowest scorer

lowest_score = None
lowest_scorer = None

for name, marks in marksByPeople.items():

    total = 0

    for mark in marks:
        total += mark

    if lowest_score is None or total < lowest_score:
        lowest_score = total
        lowest_scorer = name

print("Lowest scorer:", lowest_scorer)
```
