A dictionary is a collection of key–value pairs.
- Keys → must be immutable (string, number, tuple)
- Values → can be any data type
- Dictionaries are mutable (can change after creation)
- A dictionary is a collection which is ordered*, changeable and do not allow duplicates.
**As of Python version 3.7, dictionaries are ordered. In Python 3.6 and earlier, dictionaries are unordered.**

```python
"""
basic syntax
dictionary = {
    key: value,
    key: value
}
"""
student={
    "name":"Al Amin",
    "age":26,
    "department":"CSE"
}
print(student)
```

```python
# creating a dictionary

student={}
print(student)

# you can also use
student=dict()
```

```python
# accessing dictionary values
student={
    "name":"Al Amin",
    "age":26,
    "department":"CSE"
}

print(student["name"])
# print(student["name"]) shows KeyError if key doesn't exist
print(student['age'])

# another way -- using get()
# The difference becomes important when the key doesn't exist
print(student.get("department")) 
print(student.get('dept'))# if key doesn't exist output-- None
print(student.get('departments', 0)) # If key dont't extst i can set 0, {} etc what I want
```
